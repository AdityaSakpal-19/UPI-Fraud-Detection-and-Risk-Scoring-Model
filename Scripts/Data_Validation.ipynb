Data Cleaning and Validation

```python
import pandas as pd
df_transactions = pd.read_csv('D:/New Resumes/Latest/Projects/UPI Fraud Detection System/Data/upi_transactions.csv')
df_merchants = pd.read_csv('D:/New Resumes/Latest/Projects/UPI Fraud Detection System/Data/upi_merchants.csv')

df_transactions.info()
df_merchants.info()
```

```python
#Change data type for timestamp
df_transactions['timestamp'] = pd.to_datetime(df_transactions['timestamp'])
df_transactions.info()
```

```python
#Imputing missing values in receiver_customer_id with 'Unknown'
df_transactions['receiver_customer_id'] = df_transactions['receiver_customer_id'].fillna('Unknown')

#Encoding categorical variables
df_transactions['receiver_type'] = df_transactions['receiver_type'].map({'customer': 0, 'merchant': 1})
df_transactions['status'] = df_transactions['status'].map({'success': 1, 'failed': 0, 'pending': 0})

df_transactions = df_transactions.sort_values('timestamp')
df_transactions.head(10)
```

```python
df_transactions.drop('is_fraud', axis = 1, inplace = True)
```

```python
df_transactions.info()
```

Feature Engineering

```python
# Transaction Count in Last 24 Hours (Velocity)
from datetime import timedelta

df_transactions['txn_count_24h'] = df_transactions.groupby('sender_customer_id').apply(
    lambda group: group['timestamp'].apply(
        lambda ts: ((group['timestamp'] >= ts - timedelta(hours=24)) & 
                    (group['timestamp'] <= ts)).sum()
    )
).reset_index(level=0, drop=True)

# Transaction Count in Last 1 Hour (Velocity)
df_transactions['txn_count_1h'] = df_transactions.groupby('sender_customer_id').apply(
    lambda group: group['timestamp'].apply(
        lambda ts: ((group['timestamp'] >= ts - timedelta(hours=1)) & 
                    (group['timestamp'] <= ts)).sum()
    )
).reset_index(level=0, drop=True)
```

```python
#avg_amount_7d = Average Transaction Amount in Last 7 Days (Monetary)
df_transactions['avg_amount_7d'] = (
    df_transactions.groupby('sender_customer_id')['amount']
    .transform(lambda x: x.rolling(7).mean())
)

#high_amount_flag = High Transaction Amount Flag (Monetary)
df_transactions['high_amount_flag'] = (df_transactions['amount'] > 50000).astype(int)
```

```python
#night_txn_flag = Night Transaction Flag (Temporal)
df_transactions['hour'] = df_transactions['timestamp'].dt.hour
df_transactions['night_txn_flag'] = df_transactions['hour'].apply(lambda x: 1 if x < 6 else 0)

#collect_request_flag = Collect Request Transaction Flag (Behavioral)
df_transactions['collect_request_flag'] = (df_transactions['transaction_type'] == 'collect_request').astype(int)
```

```python
# Unique Receivers (proxy for mule detection)
unique_receivers = []
seen_receivers = {}

for _, row in df_transactions.iterrows():
    sender_id = row['sender_customer_id']
    receiver_vpa = row['receiver_vpa']
    
    if sender_id not in seen_receivers:
        seen_receivers[sender_id] = set()
    
    seen_receivers[sender_id].add(receiver_vpa)
    unique_receivers.append(len(seen_receivers[sender_id]))

df_transactions['unique_receivers'] = unique_receivers
```

```python
# Combine all signals with weighted scoring

df_transactions['fraud_score'] = 0

# Signal 1: High amount + Night transaction (high confidence)
df_transactions['fraud_score'] += (
    (df_transactions['high_amount_flag'] == 1) & 
    (df_transactions['night_txn_flag'] == 1)
).astype(int) * 3  # Weight: 3

# Signal 2: Mule detection (medium confidence)
df_transactions['fraud_score'] += (
    df_transactions['unique_receivers'] > 5
).astype(int) * 2  # Weight: 2

# Signal 3: Velocity (rapid transactions)
df_transactions['fraud_score'] += (
    df_transactions['txn_count_24h'] > 20
).astype(int) * 2  # Weight: 2

# Signal 4: Collect request flag (suspicious)
df_transactions['fraud_score'] += (
    df_transactions['collect_request_flag'] == 1
).astype(int) * 1  # Weight: 1

# Threshold: Mark as fraud if score >= 3
df_transactions['is_fraud'] = (df_transactions['fraud_score'] >= 3).astype(int)

print(df_transactions[['fraud_score', 'is_fraud']].value_counts())
```

```python
df_transactions = df_transactions.fillna(0)
df_transactions.info()
```
